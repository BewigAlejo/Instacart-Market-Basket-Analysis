- poseen `product_id` y `product_name` válidos;
- forman parte legítima del catálogo.

Por lo tanto, **no deben eliminarse ni imputarse**.

Se conservarán como una categoría propia y se tendrá en cuenta esta limitación en los análisis posteriores por pasillo o departamento.
